```python
with ZipFile(zip_path, "r") as archive:
    csv_files = sorted(
        name for name in archive.namelist()
        if name.endswith(".csv")
    )

    dataframes = [
        pd.read_csv(
            archive.open(filename),
            dtype={
                "start_station_id": "string",
                "end_station_id": "string",
            },
        )
        for filename in csv_files
    ]

df = pd.concat(dataframes, ignore_index=True)

print("Total rows:", len(df))
print("Start station ID type:", df["start_station_id"].dtype)
print("End station ID type:", df["end_station_id"].dtype)
print("Duplicate ride IDs:", df["ride_id"].duplicated().sum())
```

```python
# Convert timestamps to proper datetime types
df["started_at"] = pd.to_datetime(df["started_at"], errors="coerce")
df["ended_at"] = pd.to_datetime(df["ended_at"], errors="coerce")

# Calculate trip duration in minutes
df["duration_min"] = (
    df["ended_at"] - df["started_at"]
).dt.total_seconds() / 60

# Summarize timestamp quality
print("Invalid start timestamps:", df["started_at"].isna().sum())
print("Invalid end timestamps:", df["ended_at"].isna().sum())
print("Trips ending before they start:", (df["duration_min"] < 0).sum())
print("Zero-duration trips:", (df["duration_min"] == 0).sum())

print("\nTrip duration summary (minutes):")
print(df["duration_min"].describe())
```

```python
# Count trips exceeding selected duration thresholds
thresholds = [30, 60, 180, 720, 1440]

for minutes in thresholds:
    count = (df["duration_min"] > minutes).sum()
    percentage = count / len(df) * 100

    print(
        f"Trips over {minutes:4} minutes: "
        f"{count:>7,} ({percentage:.3f}%)"
    )

# Inspect the 10 longest trips
longest_trips = (
    df.nlargest(10, "duration_min")[
        [
            "ride_id",
            "started_at",
            "ended_at",
            "duration_min",
            "member_casual",
            "rideable_type",
        ]
    ]
)

display(longest_trips)
```

```python
# Examine trips lasting between 24 and 26 hours
long_trips = df[
    df["duration_min"].between(1440, 1560)
].copy()

print("Trips lasting 24–26 hours:", len(long_trips))

print("\nDuration statistics:")
print(long_trips["duration_min"].describe())

print("\nRider type:")
print(long_trips["member_casual"].value_counts())

print("\nBike type:")
print(long_trips["rideable_type"].value_counts())

print("\nTrips by start date:")
print(long_trips["started_at"].dt.date.value_counts().sort_index().head(15))
```

```python
# Examine the most common durations among trips over 12 hours
long_trips = df[df["duration_min"] > 720].copy()

long_trips["duration_rounded"] = (
    long_trips["duration_min"].round(0).astype(int)
)

print("Trips over 12 hours:", len(long_trips))

print("\nMost common rounded durations (minutes):")
print(
    long_trips["duration_rounded"]
    .value_counts()
    .head(15)
)

print("\nCounts by duration range:")
bins = [720, 1000, 1200, 1400, 1440, 1500, 1560]
labels = [
    "12–16.7 hours",
    "16.7–20 hours",
    "20–23.3 hours",
    "23.3–24 hours",
    "24–25 hours",
    "25–26 hours",
]

print(
    pd.cut(
        long_trips["duration_min"],
        bins=bins,
        labels=labels,
        right=False,
    ).value_counts().sort_index()
)
```

```python
# Inspect trips lasting approximately 25 hours
near_25_hours = df[
    df["duration_min"].between(1499, 1501)
].copy()

print("Trips lasting approximately 25 hours:", len(near_25_hours))

display(
    near_25_hours[
        [
            "ride_id",
            "started_at",
            "ended_at",
            "duration_min",
            "rideable_type",
            "member_casual",
        ]
    ].head(15)
)
```

```python
# Compare the suspicious trips against exactly 25 hours
near_25_hours["difference_from_25h_min"] = (
    near_25_hours["duration_min"] - (25 * 60)
)

print("Difference from exactly 25 hours (minutes):")
display(
    near_25_hours["difference_from_25h_min"].describe()
)

print("\nTrips within 1 minute of exactly 25 hours:")
print(
    near_25_hours["difference_from_25h_min"]
    .abs()
    .le(1)
    .sum()
)
```

```python
# Count approximately 25-hour trips by bike type and rider type
near_25_hours.pivot_table(
    index="rideable_type",
    columns="member_casual",
    values="ride_id",
    aggfunc="count",
    fill_value=0,
    observed=False
)
```

```python
# Flag unusual durations without deleting any records

df["duration_flag"] = "Normal"

df.loc[
    df["duration_min"] > 180,
    "duration_flag"
] = "Long: over 3 hours"

df.loc[
    df["duration_min"] > 720,
    "duration_flag"
] = "Very long: over 12 hours"

df.loc[
    df["duration_min"] >= 1440,
    "duration_flag"
] = "Investigate: 24+ hours"

# Summarize the flags
print(df["duration_flag"].value_counts())

print(
    "\nPercentage of all trips in each category:"
)

print(
    (df["duration_flag"].value_counts(normalize=True) * 100)
    .round(3)
)
```

```python
# Keep 24+ hour journeys separate for later investigation
df_24h_plus = df[df["duration_min"] >= 1440].copy()

# Main dataset: exclude 24+ hour journeys
df_main = df[df["duration_min"] < 1440].copy()

print("Original dataset:", len(df))
print("Main analysis dataset:", len(df_main))
print("Set aside for investigation:", len(df_24h_plus))

print(
    "\nRows accounted for:",
    len(df_main) + len(df_24h_plus) == len(df)
)
```
