```python
with ZipFile(zip_path, "r") as archive:
    csv_files = sorted(
        name for name in archive.namelist()
        if name.endswith(".csv")
    )

    dataframes = [
        pd.read_csv(
            archive.open(filename),
            dtype={
                "start_station_id": "string",
                "end_station_id": "string",
            },
        )
        for filename in csv_files
    ]

df = pd.concat(dataframes, ignore_index=True)

print("Total rows:", len(df))
print("Start station ID type:", df["start_station_id"].dtype)
print("End station ID type:", df["end_station_id"].dtype)
print("Duplicate ride IDs:", df["ride_id"].duplicated().sum())
```

```python
# Convert timestamps to proper datetime types
df["started_at"] = pd.to_datetime(df["started_at"], errors="coerce")
df["ended_at"] = pd.to_datetime(df["ended_at"], errors="coerce")

# Calculate trip duration in minutes
df["duration_min"] = (
    df["ended_at"] - df["started_at"]
).dt.total_seconds() / 60

# Summarize timestamp quality
print("Invalid start timestamps:", df["started_at"].isna().sum())
print("Invalid end timestamps:", df["ended_at"].isna().sum())
print("Trips ending before they start:", (df["duration_min"] < 0).sum())
print("Zero-duration trips:", (df["duration_min"] == 0).sum())

print("\nTrip duration summary (minutes):")
print(df["duration_min"].describe())
```

```python
# Count trips exceeding selected duration thresholds
thresholds = [30, 60, 180, 720, 1440]

for minutes in thresholds:
    count = (df["duration_min"] > minutes).sum()
    percentage = count / len(df) * 100

    print(
        f"Trips over {minutes:4} minutes: "
        f"{count:>7,} ({percentage:.3f}%)"
    )

# Inspect the 10 longest trips
longest_trips = (
    df.nlargest(10, "duration_min")[
        [
            "ride_id",
            "started_at",
            "ended_at",
            "duration_min",
            "member_casual",
            "rideable_type",
        ]
    ]
)

display(longest_trips)
```

## EXAMINING THE LONGER TRIPS

```python
# Examine trips lasting between 24 and 26 hours
long_trips = df[
    df["duration_min"].between(1440, 1560)
].copy()

print("Trips lasting 24–26 hours:", len(long_trips))

print("\nDuration statistics:")
print(long_trips["duration_min"].describe())

print("\nRider type:")
print(long_trips["member_casual"].value_counts())

print("\nBike type:")
print(long_trips["rideable_type"].value_counts())

print("\nTrips by start date:")
print(long_trips["started_at"].dt.date.value_counts().sort_index().head(15))
```

```python
# Examine the most common durations among trips over 12 hours
long_trips = df[df["duration_min"] > 720].copy()

long_trips["duration_rounded"] = (
    long_trips["duration_min"].round(0).astype(int)
)

print("Trips over 12 hours:", len(long_trips))

print("\nMost common rounded durations (minutes):")
print(
    long_trips["duration_rounded"]
    .value_counts()
    .head(15)
)

print("\nCounts by duration range:")
bins = [720, 1000, 1200, 1400, 1440, 1500, 1560]
labels = [
    "12–16.7 hours",
    "16.7–20 hours",
    "20–23.3 hours",
    "23.3–24 hours",
    "24–25 hours",
    "25–26 hours",
]

print(
    pd.cut(
        long_trips["duration_min"],
        bins=bins,
        labels=labels,
        right=False,
    ).value_counts().sort_index()
)
```

```python
# Inspect trips lasting approximately 25 hours
near_25_hours = df[
    df["duration_min"].between(1499, 1501)
].copy()

print("Trips lasting approximately 25 hours:", len(near_25_hours))

display(
    near_25_hours[
        [
            "ride_id",
            "started_at",
            "ended_at",
            "duration_min",
            "rideable_type",
            "member_casual",
        ]
    ].head(15)
)
```

```python
# Compare the suspicious trips against exactly 25 hours
near_25_hours["difference_from_25h_min"] = (
    near_25_hours["duration_min"] - (25 * 60)
)

print("Difference from exactly 25 hours (minutes):")
display(
    near_25_hours["difference_from_25h_min"].describe()
)

print("\nTrips within 1 minute of exactly 25 hours:")
print(
    near_25_hours["difference_from_25h_min"]
    .abs()
    .le(1)
    .sum()
)
```

```python
# Flag unusual durations without deleting any records

df["duration_flag"] = "Normal"

df.loc[
    df["duration_min"] > 180,
    "duration_flag"
] = "Long: over 3 hours"

df.loc[
    df["duration_min"] > 720,
    "duration_flag"
] = "Very long: over 12 hours"

df.loc[
    df["duration_min"] >= 1440,
    "duration_flag"
] = "Investigate: 24+ hours"

# Summarize the flags
print(df["duration_flag"].value_counts())

print(
    "\nPercentage of all trips in each category:"
)

print(
    (df["duration_flag"].value_counts(normalize=True) * 100)
    .round(3)
)
```

```python
# Keep 24+ hour journeys separate for later investigation
df_24h_plus = df[df["duration_min"] >= 1440].copy()

# Main dataset: exclude 24+ hour journeys
df_main = df[df["duration_min"] < 1440].copy()

print("Original dataset:", len(df))
print("Main analysis dataset:", len(df_main))
print("Set aside for investigation:", len(df_24h_plus))

print(
    "\nRows accounted for:",
    len(df_main) + len(df_24h_plus) == len(df)
)
```

## ANALYZING MISSING DATA

```python
# Check missing values in the main analysis dataset
missing = df_main.isna().sum()

missing_summary = pd.DataFrame({
    "missing_count": missing,
    "missing_percent": (missing / len(df_main) * 100).round(3)
})

display(
    missing_summary[missing_summary["missing_count"] > 0]
    .sort_values("missing_count", ascending=False)
)
```

```python
# Check whether missing station fields occur together
station_missing = pd.Series({
    "Missing start station name": df_main["start_station_name"].isna().sum(),
    "Missing start station ID": df_main["start_station_id"].isna().sum(),
    "Missing either start station name or ID": (
        df_main["start_station_name"].isna()
        | df_main["start_station_id"].isna()
    ).sum(),
    "Missing end station name": df_main["end_station_name"].isna().sum(),
    "Missing end station ID": df_main["end_station_id"].isna().sum(),
    "Missing either end station name or ID": (
        df_main["end_station_name"].isna()
        | df_main["end_station_id"].isna()
    ).sum(),
})

display(station_missing.to_frame("trip_count"))
```

```python
# Identify trips with missing station names
missing_start = df_main["start_station_name"].isna()
missing_end = df_main["end_station_name"].isna()

# Count each combination
overlap_summary = pd.Series({
    "Both start and end missing": (missing_start & missing_end).sum(),
    "Only start missing": (missing_start & ~missing_end).sum(),
    "Only end missing": (~missing_start & missing_end).sum(),
    "Neither missing": (~missing_start & ~missing_end).sum(),
})

display(overlap_summary.to_frame("trip_count"))

print("\nTotal trips checked:", len(df_main))
print("Counts add up correctly:", overlap_summary.sum() == len(df_main))
```

```python
# Check whether missing station names align with missing coordinates
coordinate_missing_summary = pd.Series({
    "Start name and coordinates all missing": (
        df_main["start_station_name"].isna()
        & df_main["start_lat"].isna()
        & df_main["start_lng"].isna()
    ).sum(),

    "End name and coordinates all missing": (
        df_main["end_station_name"].isna()
        & df_main["end_lat"].isna()
        & df_main["end_lng"].isna()
    ).sum(),

    "Start station missing but coordinates present": (
        df_main["start_station_name"].isna()
        & df_main["start_lat"].notna()
        & df_main["start_lng"].notna()
    ).sum(),

    "End station missing but coordinates present": (
        df_main["end_station_name"].isna()
        & df_main["end_lat"].notna()
        & df_main["end_lng"].notna()
    ).sum(),
})

display(coordinate_missing_summary.to_frame("trip_count"))
```

```python
# Flag trips with missing station information
df_main["missing_start_station"] = (
    df_main["start_station_name"].isna()
)

df_main["missing_end_station"] = (
    df_main["end_station_name"].isna()
)

# Verify the flags
print("Missing start station:", df_main["missing_start_station"].sum())
print("Missing end station:", df_main["missing_end_station"].sum())

print("\nUpdated dataset shape:", df_main.shape)
```

## Checking For Station Names & IDs Concistency

```python
# Combine start and end station information
start_stations = df_main[
    ["start_station_id", "start_station_name"]
].rename(columns={
    "start_station_id": "station_id",
    "start_station_name": "station_name"
})

end_stations = df_main[
    ["end_station_id", "end_station_name"]
].rename(columns={
    "end_station_id": "station_id",
    "end_station_name": "station_name"
})

# Combine both into one station reference table
stations = pd.concat(
    [start_stations, end_stations],
    ignore_index=True
).dropna().drop_duplicates()

# Count distinct names associated with each station ID
names_per_id = (
    stations.groupby("station_id")["station_name"]
    .nunique()
    .sort_values(ascending=False)
)

print("Unique station IDs:", stations["station_id"].nunique())
print("Unique station names:", stations["station_name"].nunique())

print(
    "Station IDs associated with multiple names:",
    (names_per_id > 1).sum()
)

display(names_per_id.head(10))
```

```python
# Count unique station IDs associated with each station name
ids_per_name = (
    stations.groupby("station_name")["station_id"]
    .nunique()
    .sort_values(ascending=False)
)

print(
    "Station names associated with multiple IDs:",
    (ids_per_name > 1).sum()
)

print("\nMaximum IDs associated with one station name:")
print(ids_per_name.max())

print("\nExamples:")
display(ids_per_name[ids_per_name > 1].head(15))
```

```python
# Select station names associated with multiple IDs
duplicate_names = ids_per_name[ids_per_name > 1].index

# Show the corresponding station IDs
station_id_pairs = (
    stations[
        stations["station_name"].isin(duplicate_names)
    ]
    .sort_values(["station_name", "station_id"])
)

display(station_id_pairs.head(30))
```

```python
# Check whether duplicate station IDs differ only by trailing zeros

station_id_pairs_check = station_id_pairs.copy()

station_id_pairs_check["normalized_id"] = (
    station_id_pairs_check["station_id"]
    .str.replace(r"(\.\d*?)0+$", r"\1", regex=True)
    .str.rstrip(".")
)

# Count distinct normalized IDs per station name
normalized_counts = (
    station_id_pairs_check.groupby("station_name")["normalized_id"]
    .nunique()
)

print("Names with equivalent IDs after normalization:")
print((normalized_counts == 1).sum())

print("\nNames still associated with different IDs:")
print((normalized_counts > 1).sum())

display(
    station_id_pairs_check[
        ["station_name", "station_id", "normalized_id"]
    ].head(20)
)
```

```python
# Identify station names whose IDs remain different
unresolved_names = normalized_counts[
    normalized_counts > 1
].index

# Display their original and normalized IDs
unresolved_stations = station_id_pairs_check[
    station_id_pairs_check["station_name"].isin(unresolved_names)
].sort_values("station_name")

display(
    unresolved_stations[
        ["station_name", "station_id", "normalized_id"]
    ]
)
```

```python
# Combine station IDs, names, and coordinates from both trip endpoints
start_locations = df_main[
    ["start_station_id", "start_station_name", "start_lat", "start_lng"]
].copy()

start_locations.columns = [
    "station_id", "station_name", "lat", "lng"
]

end_locations = df_main[
    ["end_station_id", "end_station_name", "end_lat", "end_lng"]
].copy()

end_locations.columns = [
    "station_id", "station_name", "lat", "lng"
]

all_locations = pd.concat(
    [start_locations, end_locations],
    ignore_index=True
)

# Focus on the three unresolved station names
unresolved_locations = all_locations[
    all_locations["station_name"].isin(unresolved_names)
]

# Calculate representative coordinates for each station ID
location_summary = (
    unresolved_locations
    .groupby(["station_name", "station_id"])
    .agg(
        median_lat=("lat", "median"),
        median_lng=("lng", "median"),
        observations=("lat", "count")
    )
    .reset_index()
)

display(location_summary)
```

```python
# Count how often each ID appears as a start station
start_counts = (
    df_main[
        df_main["start_station_name"].isin(unresolved_names)
    ]
    .groupby(["start_station_name", "start_station_id"])
    .size()
    .rename("start_trips")
)

# Count how often each ID appears as an end station
end_counts = (
    df_main[
        df_main["end_station_name"].isin(unresolved_names)
    ]
    .groupby(["end_station_name", "end_station_id"])
    .size()
    .rename("end_trips")
)

# Standardize the index names so they can be combined
start_counts.index = start_counts.index.set_names(
    ["station_name", "station_id"]
)

end_counts.index = end_counts.index.set_names(
    ["station_name", "station_id"]
)

# Combine counts
station_usage = pd.concat(
    [start_counts, end_counts],
    axis=1
).fillna(0).astype(int)

station_usage["total_trips"] = (
    station_usage["start_trips"] + station_usage["end_trips"]
)

display(station_usage.reset_index())
```
